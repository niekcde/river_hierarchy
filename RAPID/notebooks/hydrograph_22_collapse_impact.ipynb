# Why the segmented collapse has a modest impact in 22

This review first explains the simulated parent–child change. Slope remains a hypothesis; no predictor changes or model fits are made here.

## 1. What changed between the old and new centerlines?

The old uniform centerline advances the finite-window centroid by 11.58 hours; the segmented centerline advances it by 5.26 hours. Their mean route-time diagnostics are 19.85 and 27.49 hours, compared with 34.62 hours for the parent. These route times are not themselves hydrograph centroid shifts.

The old uniform representation travels at the 5 m/s cap throughout. The segmented representation retains actual local sections. Relative to routing those same segmented lengths at 5 m/s:

- Unchanged reaches add 4.648 hours of route time.
- Collapsed blocks with locally retained hydraulics add 2.984 hours.

Thus about 61% of this speed-related difference from the old uniform representation is on unchanged reaches. A small total-length difference adds another approximately 0.008 hours when comparing against the actual old uniform length. This decomposition concerns route time, not an exact decomposition of the 6.32-hour difference between finite-window centroid shifts.

## 2. Why is the actual parent–child change modest?

Unchanged reaches contribute 9.897 hours in **both** parent and segmented child, so they cancel in the mean route-time difference. Inside the selected blocks, mean route time changes from 24.725 to 17.594 hours: a 7.131-hour reduction, or 28.8%.

Most of the edited length is concentrated in two blocks:

| Block | Parent expected length | Parent effective speed | Child speed | Route-time reduction | Controlled centroid advance |
|---|---:|---:|---:|---:|---:|
| CB002_bubble_11 | 102.57 km | 4.05 m/s | 5.00 m/s | 1.80 h | 1.39 h |
| CB001_bubble_9 | 91.68 km | 4.59 m/s | 5.00 m/s | 1.00 h | 0.77 h |

These account for **67% of the edited expected travelled length**, but their parent routing is already close to the capped child speed. Their combined 12.59-hour route time only falls to 9.78 hours. Consequently, their large geometric size does not translate into a large change in hours.

The other blocks contribute the remaining 4.33 hours of route-time reduction. This is consistent with the existing cap sensitivity: relaxing the cap increases the centroid advance to 7.86–8.24 hours, still below the forest's 13.64-hour prediction. Those sensitivities retain calibration and are not replacement calibrated results.

## 3. New controlled check: collapse each block separately

Fourteen fresh simulations replace one complete closed block at a time while preserving every outside physical parameter exactly. They use the same forcing and published shared 0.5% window. The common baseline remains exactly the published baseline. All route successfully with nonnegative native discharge.

Sum of individual centroid shifts: **−5.19952 hours**. Combined segmented shift: **−5.26339 hours**. The difference is **−3.83 minutes** (about 1.2% of the combined shift).

Therefore a dramatic hidden interaction among simultaneously collapsed blocks is not needed to explain the modest result. The small total impact is largely visible in the individual controlled responses. Centroid shifts are not mathematically additive in general; this is an empirical finding for this event/window.


```python
from pathlib import Path
import pandas as pd
import plotly.express as px
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'RAPID').is_dir())
BASE=ROOT/'RAPID/outputs/hydrograph_closed_unit_extension_20261019'
OUT=BASE/'impact22'
controls=pd.read_csv(OUT/'controlled_block_responses.csv')
blocks=pd.read_csv(BASE/'segmented_diagnostics/block_attribution.csv',dtype={'example':str})
z=blocks[blocks.example.eq('22')].merge(controls,on='state',validate='one_to_one')
display(z[['state','parent_length_km','parent_time_hours','child_time_hours','delta_hours','centroid_shift_hours']].round(3))
print('Individual shifts summed:',z.centroid_shift_hours.sum(),'hours')
print('Combined shift:',controls.set_index('state').loc['all_closed_units_collapsed','centroid_shift_hours'],'hours')
px.bar(z.sort_values('centroid_shift_hours'),x='state',y='centroid_shift_hours',
       title='22: each block collapsed alone; all outside parameters preserved',
       labels={'centroid_shift_hours':'Centroid shift (hours)'}).show()
```

## What this establishes before revisiting the forest

The important distinction is **large geometry versus a large removable routing delay**. In 22, the long dominant blocks already route quickly before collapse, and the child is capped. Segmentation also preserves slow local sections that the old uniform representation homogenized. These are consistent parts of the explanation, but refer to different comparisons.
